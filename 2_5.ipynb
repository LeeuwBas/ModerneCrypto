```python
import random

import matplotlib.pyplot as plt


def BirthdayProb(q, N):
    prob = random.choices(range(N), k=q)
    prob = len(prob) - len(set(prob))
    return prob > 0


def BirthdayProbSim(q, N, num_trials):
    count = 0
    for _ in range(num_trials):
        if BirthdayProb(q, N):
            count += 1
    return count / num_trials


def plot_birthday_prob(N, num_trials):
    qs = list(range(1, int((2 * N) ** 0.5) + 1))
    probs = [
        [(0.632 * q * (q - 1)) / (2 * N) for q in qs],
        [BirthdayProbSim(q, N, num_trials) for q in qs],
        [(q * (q - 1)) / (2 * N) for q in qs]
    ]

    for prob in probs:
        plt.plot(qs, prob)

    plt.title(f'Birthday Problem Simulation (N={N})')
    plt.xlabel('Number of People (q)')
    plt.ylabel('Probability of Shared Birthday')
    plt.legend(['Lower bound', 'Simulation', 'Upper bound'])
    plt.grid()
    plt.show()


def plot_days_prob(q, max_N, num_trials):
    Ns = list(range(q ** 2 // 2, max_N + 1))
    probs = [
        [(0.632 * q * (q - 1)) / (2 * N) for N in Ns],
        [BirthdayProbSim(q, N, num_trials) for N in Ns],
        [(q * (q - 1)) / (2 * N) for N in Ns]
    ]

    for prob in probs:
        plt.plot(Ns, prob)

    plt.title(f'Birthday Problem Simulation (q={q})')
    plt.xlabel('Number of Days (N)')
    plt.ylabel('Probability of Shared Birthday')
    plt.legend(['Lower bound', 'Simulation', 'Upper bound'])
    plt.grid()
    plt.show()


plot_birthday_prob(700, 10_000)
plot_days_prob(20, 700, 10_000)
```
